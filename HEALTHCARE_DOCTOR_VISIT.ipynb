# univariate analysis

```python
sns.countplot(data=df, x="visits")
plt.title("Distribution of Doctor Visits")
plt.xlabel("Number of Doctor Visits")
plt.ylabel("Number of People")
plt.show()
```

```python
df.groupby("gender")["visits"].mean()
```

```python
sns.boxplot(data=df, x="gender", y="visits")
plt.title("Doctor Visits by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Doctor Visits")
plt.show()
```

```python
pd.crosstab(df["gender"], df["visits"])
```

```python
visit_gender_pct = pd.crosstab(
    df["gender"],
    df["visits"],
    normalize="index"
) * 100
```

```python
visit_gender_pct.T.plot(kind="bar")
plt.title("Doctor Visit Distribution by Gender")
plt.xlabel("Number of Doctor Visits")
plt.ylabel("Percentage of People")
plt.legend(title="Gender")
plt.show()
```

```python
df["age_years"] = df["age"] * 100
```

```python
df["age_years"].describe()
```

```python
sns.histplot(data=df, x="age_years", bins=10)

plt.title("Age Distribution")
plt.xlabel("Age (Years)")
plt.ylabel("Number of People")

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="age_years",
    y="visits",
    hue="gender"
)

plt.title("Age vs Doctor Visits by Gender")
plt.xlabel("Age (Years)")
plt.ylabel("Number of Doctor Visits")
plt.legend(title="Gender")

plt.show()
```

```python
sns.countplot(data=df, x="illness")

plt.title("Distribution of Illness")
plt.xlabel("Number of Illnesses")
plt.ylabel("Number of People")

plt.show()
```

```python
df.groupby("illness")["visits"].mean()
```

```python
illness_visits = df.groupby("illness")["visits"].mean()

illness_visits.plot(kind="bar")

plt.title("Average Doctor Visits by Number of Illnesses")
plt.xlabel("Number of Illnesses")
plt.ylabel("Average Doctor Visits")

plt.show()
```

```python
illness_analysis = df.groupby("illness")["visits"].agg(
    people="count",
    average_visits="mean"
)

illness_analysis
```

```python
correlation_matrix = df[
    ["visits", "age", "income", "illness", "reduced", "health"]
].corr()

correlation_matrix
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

## Limitations of the Analysis
- The number of observations differs across illness levels, with fewer observations in the higher-illness groups. Therefore, averages for these groups should be interpreted with appropriate caution.

- Correlation analysis measures linear relationships only and may not capture non-linear relationships between variables.

- The analysis considers selected variables available in the dataset. Other factors that may influence doctor visits may not be included.

- Some variables are represented as scores or counts, and their numerical values should be interpreted according to their original definitions rather than simply as continuous measurements.

## Key Findings and Insights

### 1. Doctor Visit Distribution
Doctor visits are highly concentrated at zero. Out of 5,190 observations, 4,141 recorded zero doctor visits, while substantially fewer observations reported one or more visits.

### 2. Gender and Doctor Visits
The distribution of doctor visits differs by gender. Females had a lower proportion of zero visits (75.57%) than males (84.36%), while females had higher proportions at one and two visits.

### 3. Age and Doctor Visits
The scatter plot does not show a clear linear relationship between age and doctor visits. Doctor visits occur across different age groups, with no obvious consistent upward or downward pattern.

### 4. Illness and Doctor Visits
Average doctor visits generally increased with the number of recorded illnesses, rising from 0.079 visits for people with zero illnesses to 0.814 visits for people with five illnesses.

### 5. Reduced Activity and Doctor Visits
The number of days with reduced normal activity had the strongest positive linear association with doctor visits among the numerical variables examined, with a correlation of 0.42.

## Executive Summary

This project analyzes a healthcare dataset containing 5,190 observations and 13 variables to explore patterns in doctor visits and their relationship with demographic, socioeconomic, and health-related variables.

The analysis shows that doctor visits are highly concentrated at zero, with 4,141 observations reporting no visits. The distribution of visits also differs by gender, with females showing a lower proportion of zero visits and higher proportions of one and two visits compared with males.

Age was examined through its distribution and its relationship with doctor visits. The data does not show a clear linear relationship between age and the number of visits.

A notable pattern was observed between illness count and doctor visits: average visits generally increased as the number of recorded illnesses increased. Correlation analysis further showed that the number of days with reduced normal activity had the strongest positive linear association with doctor visits among the numerical variables examined (r = 0.42).

Overall, the analysis identifies several associations between health-related characteristics and doctor visits. However, these findings are observational and should not be interpreted as evidence of causation.

## Conclusion

This analysis explored the distribution of doctor visits and examined their relationships with demographic, socioeconomic, and health-related variables.

The analysis found that doctor visits are highly concentrated at zero, while visit patterns differ across gender groups. Age did not show a clear linear relationship with doctor visits in the scatter plot. In contrast, average doctor visits generally increased with the number of recorded illnesses. The correlation analysis also identified a positive linear association between days of reduced normal activity and doctor visits (r = 0.42).

Overall, the findings suggest that health-related characteristics, particularly illness count and reduced normal activity, are more strongly associated with doctor visits than some of the demographic and socioeconomic variables examined. However, these relationships are observational and should not be interpreted as causal effects.

The analysis provides a descriptive view of healthcare utilization within this dataset and highlights relationships that could be investigated further using additional variables and more advanced statistical methods.
